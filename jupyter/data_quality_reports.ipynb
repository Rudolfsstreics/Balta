```python
import sqlalchemy
import pandas as pd
from ydata_profiling import ProfileReport
from local_secrets import username,password
```

```python
from sqlalchemy.engine import URL
connection_url = URL.create(
    "mssql+pyodbc",
    username=username,
    password=password,
    host="localhost",
    port=1433,
    database="balta_bronze",
    query={
        "driver": "ODBC Driver 17 for SQL Server",
        "TrustServerCertificate": "yes",
    },
)
# engine = sqlalchemy.create_engine('mssql://localhost/balta_bronze?trusted_connection=yes')
connection_url
engine = sqlalchemy.create_engine(connection_url)
```

```python

with engine.connect() as connection:
    for index, row in pd.read_sql("SELECT *  FROM INFORMATION_SCHEMA.TABLES WHERE TABLE_TYPE='BASE TABLE'",connection).iterrows():
        print(row['TABLE_NAME'])
        df = pd.read_sql(f"SELECT *  FROM {row['TABLE_NAME']}",connection)
        profile = ProfileReport(df, title=f"{row['TABLE_NAME']} Profiling Report")
        profile.to_file(f"profiling_results\{row['TABLE_NAME']}.html")
```

```python
with engine.connect() as connection:
    for index, row in pd.read_sql("SELECT *  FROM INFORMATION_SCHEMA.TABLES WHERE TABLE_TYPE='BASE TABLE'",connection).iterrows():
        print(row['TABLE_NAME'])
        df = pd.read_sql(f"""
            SELECT 
            C.[customer_id]
            ,C.[first_name]
            ,C.[last_name]
            ,C.[birth_date]
            ,C.[gender]
            ,C.[city]
            ,C.[segment]
            ,C.[created_at]
            ,C.[country]
            ,P.[policy_id]
            ,P.[product_code]
            ,P.[product_variant]
            ,P.[start_date]
            ,P.[end_date]
            ,P.[policy_status]
            ,P.[written_premium]
            ,P.[sum_insured]
            ,P.[currency]
            ,P.[sales_channel]
            ,P.[policy_version]
            ,CL.[claim_id]
            ,CL.[claim_date]
            ,CL.[claim_amount]
            FROM [dbo].[customers] AS C 
            FULL OUTER JOIN [balta_bronze].[dbo].[policies] as P
            ON  C.[customer_id]=P.[customer_id]
            FULL OUTER JOIN [balta_bronze].[dbo].[claims] as CL
            ON P.[policy_id] =CL.[policy_id]
        """,connection)
        profile = ProfileReport(df, title=f"customers policies claims Profiling Report")
        profile.to_file(f"profiling_results\customers policies claims Profiling Report.html")
```
